```python
# -------------------------------
# MODULER & PACKAGER
# -------------------------------

###### utilities/datetime_utils.py

from datetime import datetime, date
import time

def current_datetime():
    now = datetime.now()
    print("Current Date and Time:", now.strftime("%d-%m-%Y %I:%M:%S %p"))


def date_difference():
    try:
        date1 = datetime.strptime(
            input("Enter first date (DD-MM-YYYY): "),
            "%d-%m-%Y"
        ).date()

        date2 = datetime.strptime(
            input("Enter second date (DD-MM-YYYY): "),
            "%d-%m-%Y"
        ).date()

        print("Difference:", abs((date2 - date1).days), "days")

    except ValueError:
        print("Invalid date format!")


def custom_date_format():
    try:
        value = input("Enter date (DD-MM-YYYY): ")
        dt = datetime.strptime(value, "%d-%m-%Y")

        print("DD/MM/YYYY:", dt.strftime("%d/%m/%Y"))
        print("Month Name:", dt.strftime("%B"))
        print("Day Name:", dt.strftime("%A"))
        print("YYYY-MM-DD:", dt.strftime("%Y-%m-%d"))

    except ValueError:
        print("Invalid date format!")


def time_difference():
    try:
        start = datetime.strptime(
            input("Enter start time (HH:MM): "), "%H:%M"
        )
        end = datetime.strptime(
            input("Enter end time (HH:MM): "), "%H:%M"
        )

        difference = abs((end - start).total_seconds())
        hours, remainder = divmod(int(difference), 3600)
        minutes, seconds = divmod(remainder, 60)

        print(f"Time difference: {hours} hours, "
              f"{minutes} minutes, {seconds} seconds")

    except ValueError:
        print("Enter time in HH:MM format.")


def working_hours():
    try:
        start = datetime.strptime(
            input("Enter start time (HH:MM): "), "%H:%M"
        )
        end = datetime.strptime(
            input("Enter end time (HH:MM): "), "%H:%M"
        )

        if end < start:
            print("End time must be after start time.")
            return

        hours = (end - start).total_seconds() / 3600
        print(f"Total working hours: {hours:.2f}")

    except ValueError:
        print("Invalid time format!")


def stopwatch():
    input("Press Enter to start the stopwatch...")
    start = time.perf_counter()

    input("Press Enter to stop the stopwatch...")
    elapsed = time.perf_counter() - start

    print(f"Elapsed time: {elapsed:.2f} seconds")


def countdown_timer():
    try:
        seconds = int(input("Enter countdown seconds: "))

        if seconds < 0:
            print("Enter a non-negative number.")
            return

        while seconds > 0:
            print(f"Time remaining: {seconds} seconds")
            time.sleep(1)
            seconds -= 1

        print("Countdown finished!")

    except ValueError:
        print("Enter a valid integer.")


def datetime_menu():
    while True:
        print("\n--- Date and Time Operations ---")
        print("1. Current Date and Time")
        print("2. Difference Between Dates")
        print("3. Custom Date Formatting")
        print("4. Difference Between Times")
        print("5. Calculate Working Hours")
        print("6. Stopwatch")
        print("7. Countdown Timer")
        print("8. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1":
            current_datetime()
        elif choice == "2":
            date_difference()
        elif choice == "3":
            custom_date_format()
        elif choice == "4":
            time_difference()
        elif choice == "5":
            working_hours()
        elif choice == "6":
            stopwatch()
        elif choice == "7":
            countdown_timer()
        elif choice == "8":
            break
        else:
            print("Invalid choice!")

# utilities/math_utils.py

import math


def arithmetic():
    try:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))

        print("Addition:", a + b)
        print("Subtraction:", a - b)
        print("Multiplication:", a * b)

        if b != 0:
            print("Division:", a / b)
            print("Floor Division:", a // b)
            print("Remainder:", a % b)
        else:
            print("Division by zero is not allowed.")

        print("Power:", a ** b)

    except ValueError:
        print("Enter valid numbers.")


def trigonometry():
    try:
        angle = float(input("Enter angle in degrees: "))
        radians = math.radians(angle)

        print("Sin:", round(math.sin(radians), 4))
        print("Cos:", round(math.cos(radians), 4))
        print("Tan:", round(math.tan(radians), 4))

    except ValueError:
        print("Enter a valid angle.")


def factorial():
    try:
        n = int(input("Enter a non-negative integer: "))

        if n < 0:
            print("Factorial is not defined for negative integers.")
        else:
            print("Factorial:", math.factorial(n))

    except ValueError:
        print("Enter a valid integer.")


def logarithm():
    try:
        n = float(input("Enter a positive number: "))

        if n <= 0:
            print("Logarithm requires a positive number.")
        else:
            print("Natural Log (ln):", math.log(n))
            print("Log Base 10:", math.log10(n))
            print("Log Base 2:", math.log2(n))

    except ValueError:
        print("Enter a valid number.")


def compound_interest():
    try:
        principal = float(input("Enter principal amount: ₹"))
        rate = float(input("Enter annual interest rate (%): "))
        years = float(input("Enter time in years: "))

        if principal < 0 or rate < 0 or years < 0:
            print("Values cannot be negative.")
            return

        amount = principal * (1 + rate / 100) ** years
        interest = amount - principal

        print(f"Final Amount: ₹{amount:.2f}")
        print(f"Compound Interest: ₹{interest:.2f}")

    except ValueError:
        print("Enter valid numbers.")


def geometric_area():
    try:
        print("\n1. Circle")
        print("2. Rectangle")
        print("3. Triangle")

        choice = input("Choose shape: ")

        if choice == "1":
            radius = float(input("Enter radius: "))
            if radius < 0:
                print("Radius cannot be negative.")
                return
            print("Circle Area:", math.pi * radius ** 2)

        elif choice == "2":
            length = float(input("Enter length: "))
            width = float(input("Enter width: "))

            if length < 0 or width < 0:
                print("Dimensions cannot be negative.")
                return

            print("Rectangle Area:", length * width)

        elif choice == "3":
            base = float(input("Enter base: "))
            height = float(input("Enter height: "))

            if base < 0 or height < 0:
                print("Dimensions cannot be negative.")
                return

            print("Triangle Area:", 0.5 * base * height)

        else:
            print("Invalid shape choice.")

    except ValueError:
        print("Enter valid dimensions.")


def unit_conversion():
    try:
        print("\n1. Kilometers to Miles")
        print("2. Miles to Kilometers")
        print("3. Celsius to Fahrenheit")
        print("4. Fahrenheit to Celsius")
        print("5. Kilograms to Pounds")
        print("6. Pounds to Kilograms")

        choice = input("Choose conversion: ")
        value = float(input("Enter value: "))

        if choice == "1":
            print("Miles:", value * 0.621371)
        elif choice == "2":
            print("Kilometers:", value / 0.621371)
        elif choice == "3":
            print("Fahrenheit:", (value * 9 / 5) + 32)
        elif choice == "4":
            print("Celsius:", (value - 32) * 5 / 9)
        elif choice == "5":
            print("Pounds:", value * 2.20462)
        elif choice == "6":
            print("Kilograms:", value / 2.20462)
        else:
            print("Invalid conversion choice.")

    except ValueError:
        print("Enter a valid number.")


def math_menu():
    while True:
        print("\n--- Mathematical Operations ---")
        print("1. Arithmetic Operations")
        print("2. Trigonometry")
        print("3. Factorial")
        print("4. Logarithms")
        print("5. Compound Interest")
        print("6. Area of Geometric Shapes")
        print("7. Unit Conversion")
        print("8. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1":
            arithmetic()
        elif choice == "2":
            trigonometry()
        elif choice == "3":
            factorial()
        elif choice == "4":
            logarithm()
        elif choice == "5":
            compound_interest()
        elif choice == "6":
            geometric_area()
        elif choice == "7":
            unit_conversion()
        elif choice == "8":
            break
        else:
            print("Invalid choice!")

# utilities/random_utils.py

import random
import string


def random_number():
    try:
        start = int(input("Enter minimum number: "))
        end = int(input("Enter maximum number: "))

        if start > end:
            print("Minimum must not exceed maximum.")
            return

        print("Random Number:", random.randint(start, end))

    except ValueError:
        print("Enter valid integers.")


def random_list():
    try:
        count = int(input("How many random numbers? "))
        start = int(input("Enter minimum: "))
        end = int(input("Enter maximum: "))

        if count < 0 or start > end:
            print("Invalid range or count.")
            return

        numbers = [
            random.randint(start, end)
            for _ in range(count)
        ]

        print("Random List:", numbers)

    except ValueError:
        print("Enter valid integers.")


def generate_password():
    try:
        length = int(input("Enter password length (minimum 8): "))

        if length < 8 or length > 128:
            print("Choose a length between 8 and 128.")
            return

        characters = (
            string.ascii_letters
            + string.digits
            + string.punctuation
        )

        # secrets is more suitable for real security-sensitive passwords.
        import secrets
        password = "".join(
            secrets.choice(characters) for _ in range(length)
        )

        print("Generated Password:", password)

    except ValueError:
        print("Enter a valid length.")


def random_sampling():
    try:
        items = input(
            "Enter items separated by commas: "
        ).split(",")

        items = [item.strip() for item in items if item.strip()]

        if not items:
            print("Please enter some items.")
            return

        count = int(input("How many items should be sampled? "))

        if count < 0 or count > len(items):
            print(f"Choose between 0 and {len(items)} items.")
            return

        print("Random Sample:", random.sample(items, count))

    except ValueError:
        print("Enter a valid sample size.")


def generate_otp():
    otp = "".join(str(secrets_digit()) for _ in range(6))
    print("Demo OTP:", otp)


def secrets_digit():
    import secrets
    return secrets.randbelow(10)


def number_guessing_game():
    secret_number = random.randint(1, 100)
    attempts = 0

    print("Guess the number between 1 and 100.")

    while True:
        try:
            guess = int(input("Your guess: "))

            if not 1 <= guess <= 100:
                print("Enter a number from 1 to 100.")
                continue

            attempts += 1

            if guess < secret_number:
                print("Too low! Try again.")
            elif guess > secret_number:
                print("Too high! Try again.")
            else:
                print(f"Correct! Attempts: {attempts}")
                break

        except ValueError:
            print("Enter a valid integer.")


def random_menu():
    while True:
        print("\n--- Random Data Generation ---")
        print("1. Generate Random Number")
        print("2. Generate Random List")
        print("3. Generate Password")
        print("4. Random Sampling")
        print("5. Generate 6-Digit OTP")
        print("6. Number Guessing Game")
        print("7. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1":
            random_number()
        elif choice == "2":
            random_list()
        elif choice == "3":
            generate_password()
        elif choice == "4":
            random_sampling()
        elif choice == "5":
            generate_otp()
        elif choice == "6":
            number_guessing_game()
        elif choice == "7":
            break
        else:
            print("Invalid choice!")

# utilities/uuid_utils.py

import uuid


def generate_uuid():
    unique_id = uuid.uuid4()
    print("Generated UUID:", unique_id)


def generate_multiple_uuids():
    try:
        count = int(input("How many UUIDs? "))

        if count < 1 or count > 1000:
            print("Choose between 1 and 1000.")
            return

        for i in range(count):
            print(f"UUID {i + 1}: {uuid.uuid4()}")

    except ValueError:
        print("Enter a valid integer.")


def uuid_menu():
    while True:
        print("\n--- UUID Operations ---")
        print("1. Generate One UUID")
        print("2. Generate Multiple UUIDs")
        print("3. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1":
            generate_uuid()
        elif choice == "2":
            generate_multiple_uuids()
        elif choice == "3":
            break
        else:
            print("Invalid choice!")

# utilities/file_utils.py

from pathlib import Path

# Store project files in a dedicated folder.
DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)


def get_file_path():
    filename = input("Enter filename (example: report.txt): ").strip()

    if not filename or Path(filename).name != filename:
        print("Enter a filename without folder paths.")
        return None

    if not filename.lower().endswith(".txt"):
        print("Only .txt files are allowed.")
        return None

    return DATA_DIR / filename


def write_file():
    path = get_file_path()

    if path is None:
        return

    content = input("Enter text to save: ")

    try:
        with path.open("w", encoding="utf-8") as file:
            file.write(content + "\n")

        print("File saved successfully:", path.name)

    except OSError as error:
        print("File error:", error)


def read_file():
    path = get_file_path()

    if path is None:
        return

    if not path.is_file():
        print("File does not exist.")
        return

    try:
        with path.open("r", encoding="utf-8") as file:
            print("\n--- File Contents ---")
            print(file.read())

    except OSError as error:
        print("File error:", error)


def append_file():
    path = get_file_path()

    if path is None:
        return

    content = input("Enter text to append: ")

    try:
        with path.open("a", encoding="utf-8") as file:
            file.write(content + "\n")

        print("Text appended successfully.")

    except OSError as error:
        print("File error:", error)


def update_file():
    path = get_file_path()

    if path is None:
        return

    if not path.is_file():
        print("File does not exist.")
        return

    old_text = input("Enter text to find: ")
    new_text = input("Enter replacement text: ")

    if not old_text:
        print("Search text cannot be empty.")
        return

    try:
        content = path.read_text(encoding="utf-8")

        if old_text not in content:
            print("Text not found.")
            return

        updated_content = content.replace(old_text, new_text)

        confirm = input("Save changes? (y/n): ").lower()

        if confirm == "y":
            path.write_text(updated_content, encoding="utf-8")
            print("File updated successfully.")
        else:
            print("Update cancelled.")

    except OSError as error:
        print("File error:", error)


def file_menu():
    while True:
        print("\n--- File Operations ---")
        print("1. Write New File")
        print("2. Read File")
        print("3. Append to File")
        print("4. Find and Replace Text")
        print("5. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1":
            write_file()
        elif choice == "2":
            read_file()
        elif choice == "3":
            append_file()
        elif choice == "4":
            update_file()
        elif choice == "5":
            break
        else:
            print("Invalid choice!")

import datetime
import time
import math
import random
import uuid
import secrets
import string
from pathlib import Path

# Create a data folder
DATA_DIR = Path.cwd() / "data"
DATA_DIR.mkdir(parents=True, exist_ok=True)


# 1. DATETIME AND TIME OPERATIONS
def datetime_menu():
    while True:
        print("\n--- Date and Time Operations ---")
        print("1. Current Date and Time")
        print("2. Difference Between Dates")
        print("3. Custom Date Formatting")
        print("4. Stopwatch")
        print("5. Countdown Timer")
        print("6. Back")

        choice = input("Enter choice: ").strip()

        if choice == "1":
            print(datetime.datetime.now().strftime(
                "%d-%m-%Y %I:%M:%S %p"
            ))

        elif choice == "2":
            try:
                d1 = datetime.datetime.strptime(
                    input("First date (DD-MM-YYYY): ").strip(),
                    "%d-%m-%Y"
                ).date()

                d2 = datetime.datetime.strptime(
                    input("Second date (DD-MM-YYYY): ").strip(),
                    "%d-%m-%Y"
                ).date()

                print("Difference:", abs((d2 - d1).days), "days")

            except ValueError:
                print("Invalid date! Use DD-MM-YYYY.")

        elif choice == "3":
            try:
                d = datetime.datetime.strptime(
                    input("Enter date (DD-MM-YYYY): ").strip(),
                    "%d-%m-%Y"
                )
                print(d.strftime("%A, %d %B %Y"))

            except ValueError:
                print("Invalid date! Use DD-MM-YYYY.")

        elif choice == "4":
            input("Press Enter to start...")
            start = time.perf_counter()
            input("Press Enter to stop...")
            elapsed = time.perf_counter() - start
            print("Elapsed:", round(elapsed, 2), "seconds")

        elif choice == "5":
            try:
                seconds = int(input("Enter seconds (0-3600): "))

                if not 0 <= seconds <= 3600:
                    print("Enter a value from 0 to 3600.")
                    continue

                while seconds > 0:
                    print(seconds)
                    time.sleep(1)
                    seconds -= 1

                print("Countdown finished!")

            except ValueError:
                print("Enter a valid integer.")

        elif choice == "6":
            break

        else:
            print("Invalid choice! Select 1 to 6.")


# 2. MATHEMATICAL OPERATIONS
def math_menu():
    while True:
        print("\n--- Mathematical Operations ---")
        print("1. Arithmetic")
        print("2. Trigonometry")
        print("3. Factorial")
        print("4. Logarithm")
        print("5. Compound Interest")
        print("6. Area of Circle")
        print("7. Back")

        choice = input("Enter choice: ").strip()

        if choice == "7":
            break

        if choice not in ("1", "2", "3", "4", "5", "6"):
            print("Invalid choice! Select 1 to 7.")
            continue

        try:
            if choice == "1":
                a = float(input("First number: "))
                b = float(input("Second number: "))

                print("Addition:", a + b)
                print("Subtraction:", a - b)
                print("Multiplication:", a * b)

                if b != 0:
                    print("Division:", a / b)
                else:
                    print("Cannot divide by zero.")

            elif choice == "2":
                degrees = float(input("Angle in degrees: "))
                angle = math.radians(degrees)

                print("Sin:", round(math.sin(angle), 6))
                print("Cos:", round(math.cos(angle), 6))

                if math.isclose(math.cos(angle), 0.0, abs_tol=1e-12):
                    print("Tan: Undefined")
                else:
                    print("Tan:", round(math.tan(angle), 6))

            elif choice == "3":
                n = int(input("Enter non-negative integer: "))

                if n < 0:
                    print("Number cannot be negative.")
                elif n > 10000:
                    print("Number is too large.")
                else:
                    print("Factorial:", math.factorial(n))

            elif choice == "4":
                n = float(input("Enter positive number: "))

                if n <= 0:
                    print("Number must be positive.")
                else:
                    print("Natural log:", math.log(n))
                    print("Log base 10:", math.log10(n))

            elif choice == "5":
                p = float(input("Principal: "))
                r = float(input("Annual rate (%): "))
                t = float(input("Years: "))

                if not all(math.isfinite(x) for x in (p, r, t)):
                    print("Enter finite numbers.")
                elif p < 0 or r < 0 or t < 0:
                    print("Values cannot be negative.")
                else:
                    amount = p * (1 + r / 100) ** t
                    print("Amount:", round(amount, 2))
                    print("Interest:", round(amount - p, 2))

            elif choice == "6":
                radius = float(input("Radius: "))

                if not math.isfinite(radius):
                    print("Enter a finite number.")
                elif radius < 0:
                    print("Radius cannot be negative.")
                else:
                    print("Area:", round(math.pi * radius ** 2, 2))

        except (ValueError, OverflowError):
            print("Invalid input or result too large.")


# 3. RANDOM DATA GENERATION
def random_menu():
    while True:
        print("\n--- Random Data Generation ---")
        print("1. Random Number")
        print("2. Random List")
        print("3. Secure Random Password")
        print("4. Random Sampling")
        print("5. Six-Digit OTP")
        print("6. Back")

        choice = input("Enter choice: ").strip()

        if choice == "6":
            break

        if choice not in ("1", "2", "3", "4", "5"):
            print("Invalid choice! Select 1 to 6.")
            continue

        try:
            if choice == "1":
                a = int(input("Minimum: "))
                b = int(input("Maximum: "))

                if a > b:
                    print("Minimum cannot exceed maximum.")
                else:
                    print("Random number:", random.randint(a, b))

            elif choice == "2":
                n = int(input("How many numbers? "))

                if not 0 <= n <= 1000:
                    print("Choose between 0 and 1000.")
                else:
                    numbers = [
                        random.randint(1, 100) for _ in range(n)
                    ]
                    print("Random list:", numbers)

            elif choice == "3":
                n = int(input("Password length (8-64): "))

                if not 8 <= n <= 64:
                    print("Choose a length from 8 to 64.")
                else:
                    chars = string.ascii_letters + string.digits + "!@#$%&*"
                    password = "".join(
                        secrets.choice(chars) for _ in range(n)
                    )
                    print("Password:", password)

            elif choice == "4":
                items = input(
                    "Enter comma-separated items: "
                ).split(",")

                items = [x.strip() for x in items if x.strip()]
                n = int(input("Number of sampled items: "))

                if 0 <= n <= len(items):
                    print("Sample:", random.sample(items, n))
                else:
                    print("Sample size must be between 0 and", len(items))

            elif choice == "5":
                otp = f"{secrets.randbelow(1000000):06d}"
                print("Demo OTP:", otp)

        except ValueError:
            print("Enter valid numbers.")


# 4. UUID GENERATION
def uuid_menu():
    while True:
        print("\n--- UUID Operations ---")
        print("1. Generate One UUID")
        print("2. Generate Multiple UUIDs")
        print("3. Back")

        choice = input("Enter choice: ").strip()

        if choice == "1":
            print("UUID:", uuid.uuid4())

        elif choice == "2":
            try:
                n = int(input("How many UUIDs? "))

                if 1 <= n <= 1000:
                    for i in range(n):
                        print(f"UUID {i + 1}:", uuid.uuid4())
                else:
                    print("Choose between 1 and 1000.")

            except ValueError:
                print("Enter a valid integer.")

        elif choice == "3":
            break

        else:
            print("Invalid choice! Select 1 to 3.")


# 5. FILE OPERATIONS
def file_menu():
    while True:
        print("\n--- File Operations ---")
        print("1. Write File")
        print("2. Read File")
        print("3. Append to File")
        print("4. Find and Replace")
        print("5. Back")

        choice = input("Enter choice: ").strip()

        if choice == "5":
            break

        if choice not in ("1", "2", "3", "4"):
            print("Invalid choice! Select 1 to 5.")
            continue

        filename = input(
            "Filename (example: report.txt): "
        ).strip()

        # Accept only a simple .txt filename
        if (
            not filename.endswith(".txt")
            or Path(filename).name != filename
            or filename in (".", "..")
        ):
            print("Enter a valid .txt filename without a folder path.")
            continue

        path = DATA_DIR / filename

        try:
            if choice == "1":
                content = input("Enter text: ")
                path.write_text(content + "\n", encoding="utf-8")
                print("File saved.")

            elif choice == "2":
                if path.is_file():
                    print(path.read_text(encoding="utf-8"))
                else:
                    print("File does not exist.")

            elif choice == "3":
                content = input("Enter text to append: ")

                with path.open("a", encoding="utf-8") as f:
                    f.write(content + "\n")

                print("Text appended.")

            elif choice == "4":
                if not path.is_file():
                    print("File does not exist.")
                    continue

                old = input("Text to find: ")
                new = input("Replacement text: ")

                if not old:
                    print("Search text cannot be empty.")
                    continue

                content = path.read_text(encoding="utf-8")

                if old not in content:
                    print("Text not found.")
                elif input("Save changes? (y/n): ").strip().lower() == "y":
                    path.write_text(
                        content.replace(old, new),
                        encoding="utf-8"
                    )
                    print("File updated.")
                else:
                    print("Changes cancelled.")

        except OSError as error:
            print("File error:", error)


# 6. DYNAMIC MODULE EXPLORATION
def explore_module():
    modules = {
        "datetime": datetime,
        "time": time,
        "math": math,
        "random": random,
        "uuid": uuid,
        "secrets": secrets,
        "string": string
    }

    print("\nAvailable modules:", ", ".join(modules))
    name = input("Enter module name: ").strip()

    if name in modules:
        print(f"\nAttributes of {name}:")
        print(dir(modules[name]))
    else:
        print("Module not found.")


# MAIN MENU
def main():
    while True:
        print("\n" + "=" * 35)
        print("      MULTI-UTILITY TOOLKIT")
        print("=" * 35)
        print("1. Datetime and Time Operations")
        print("2. Mathematical Operations")
        print("3. Random Data Generation")
        print("4. Generate Unique Identifiers (UUID)")
        print("5. File Operations")
        print("6. Explore Module Attributes (dir())")
        print("7. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            datetime_menu()
        elif choice == "2":
            math_menu()
        elif choice == "3":
            random_menu()
        elif choice == "4":
            uuid_menu()
        elif choice == "5":
            file_menu()
        elif choice == "6":
            explore_module()
        elif choice == "7":
            print("Thank you for using Multi-Utility Toolkit!")
            break
        else:
            print("Invalid choice! Select 1 to 7.")


if __name__ == "__main__":
    main()
```
